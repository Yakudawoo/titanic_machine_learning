6. Separate the target variable (Y) from the explanatory variables (X)

## Preprocessing - scikit-learn part 🔬🔬
7. Separate your data to create a train set and a test set, the latter should represent 15% of the available data.

```python
# Proportion de survivants dans le dataset complet
print("Survival ratio in full dataset:")
print(Y.value_counts(normalize=True)) # normalize=True juste pour afficher les proportions
# alerte dataset déséquilibré la classe des survivants est trop faible => stratification nécessaire lors du split
# pour que train et test reproduisent exactement le déséquilibre d'origine
# condition essentielle pour comparer les performances des modèles
```

```python
from sklearn.model_selection import train_test_split

print("Dividing into train and test sets...")

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.15,   # 15% of data in test set
    random_state=42,  # ensures reproducibility
    stratify=Y        # keeps same survival ratio in train/test parce que le jeu de données est déséquilibré
)

print("...Done.")
```

8. Create the preprocessing pipeline for numeric columns

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# List of numeric columns
numeric_features = ["Age", "SibSp", "Parch", "Fare", "Pclass"]
# oneHot encoding for Pclass could be an option too, but let's keep it numeric for simplicity
# 1 riches, 2 classe moyenne, 3 pauvres

# Pipeline for numeric preprocessing
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    # 77% valeurs manquantes dans cabin, 20% dans age
    # valeurs manquantes possiblement non aléatoires 
    # pour éviter des biais d'échantillonnage, on utilise la médiane et non dropna
    # (plus robuste que la moyenne aux valeurs extrêmes)
    # conserve la distribution globale
    ("scaler", StandardScaler())
])

numeric_pipeline
```

```python
X_train.head() # see where the numeric columns are in X_train
```

9. Create the preprocessing pipeline for category columns

```python
from sklearn.preprocessing import OneHotEncoder

# List of categorical columns
categorical_features = ["Sex", "Embarked"]

# Pipeline for categorical preprocessing
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    # imputation
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

categorical_pipeline
```

10. Use the preprocessing pipelines of questions 9 and 10 to transform X_train and X_test

Reminder: you need to call `fit_transform()` on X_train and only `transform()` on X_test, to ensure that the latter gets the same transformations as X_train.

```python
from sklearn.compose import ColumnTransformer

# Regroupement des pipelines
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numeric_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)
preprocessor
```

```python
print("Performing preprocessings on train set...")
X_train_preprocessed = preprocessor.fit_transform(X_train)
print("...Done.")
# le modèle ne doit pas voir les données de test pendant l'entraînement
# il apprend les statistiques (moyenne, écart-type, etc.) uniquement sur les données d'entraînement

print("\nPerforming preprocessings on test set...")
X_test_preprocessed = preprocessor.transform(X_test)
print("...Done.")
# le test set sert uniquement à évaluer les performances du modèle après entraînement
# il doit donc être transformé avec transform() et non fit_transform()
# et subir exactement les mêmes transformations que le train set
```

```python
print("\nPreview of transformed X_train:")
print(X_train_preprocessed[:5])

print("\nPreview of transformed X_test:")
print(X_test_preprocessed[:5])
```

### Training model

```python
from sklearn.linear_model import LogisticRegression
```

```python
# Train model on PREPROCESSED data
model = LogisticRegression()

print("Training model...")
model.fit(X_train_preprocessed, Y_train)
print("...Done.")
```

### Predictions

```python
print("Predictions on training set...")
Y_train_pred = model.predict(X_train_preprocessed)
print("...Done.")
print(Y_train_pred[:5])
print()
```

```python
print("Predictions on test set...")
Y_test_pred = model.predict(X_test_preprocessed)
print("...Done.")
print(Y_test_pred[:5])
print()
```

### Performances evaluation

```python
from sklearn.metrics import accuracy_score

print("Accuracy on training set : ", accuracy_score(Y_train, Y_train_pred))
print("Accuracy on test set : ", accuracy_score(Y_test, Y_test_pred))
```
